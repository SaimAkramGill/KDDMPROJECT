#### Changing the roman names to english 
After checking the dataset finds out language difference and convert the columns to english like in the Identity Card 1 table. 

#### Also the first name and last names were in latin and finds out names ending with 'ius' and 'us'. So I have removed them and convert them to english

```python
translation_dict = {
    'numerus_internationalis_ad_identitatem': 'international_id',
    'numerus_identificationis_ad_medicinam': 'medical_id',
    'praenomen': 'first_name',
    'cognomen': 'last_name',
    'sexus': 'gender',
    'aetas_annorum': 'age',
    'urbs_natalis': 'birth_city',
    'natio': 'nationality'
}
identity_card_0.rename(columns=translation_dict, inplace=True)
```

```python
def roman_to_int(s):
    if pd.isna(s): return s
    roman_values = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    total = 0
    prev_value = 0
    # Process string in reverse
    for char in reversed(str(s).strip().upper()):
        if char not in roman_values: 
            return s # fallback if there is unexpected non-roman text
        value = roman_values[char]
        if value < prev_value:
            total -= value
        else:
            total += value
        prev_value = value
    return total
# Apply the conversion to the 'international_id' column
identity_card_0['international_id'] = identity_card_0['international_id'].apply(roman_to_int)

#Writing the names in english - converting them because have find the moms-nots.json dataset has name combine with this dataset

def strip_latin_suffix(name):
    """
    Strip common Latin suffixes from Latinized names.
    Examples:
      'Alleeius' -> 'Allee'    (strip -ius)
      'Kohlius'  -> 'Kohl'     (strip -ius)
      'Keithus'  -> 'Keith'    (strip -us)
      'Donaldus' -> 'Donald'   (strip -us)
    """
    if pd.isna(name):
        return name
    name = str(name)
    # Order matters: check -ius before -us (longer match first!)
    if name.endswith('ius'):
        return name[:-3]   # remove last 3 characters
    elif name.endswith('us'):
        return name[:-2]   # remove last 2 characters
    return name  # not Latinized, return as-is

# Apply ONLY to identity_card_0 (identity_card_1 names are already normal)
identity_card_0['first_name'] = identity_card_0['first_name'].apply(strip_latin_suffix)
identity_card_0['last_name']  = identity_card_0['last_name'].apply(strip_latin_suffix)

# Verify the result
print("After stripping Latin suffixes:")
display(identity_card_0[['first_name', 'last_name']].head(10))

print("\n--- Cleaned Identity Card 0 Data ---")
display(identity_card_0.head())
```

```python
# Step 1: Make a join key in BOTH DataFrames
# Lowercase both names and combine them into one string
# Example: "keith" + "allee" = "keith_allee"

master['name_key'] = (
    master['first_name'].str.lower().str.strip() + '_' +
    master['last_name'].str.lower().str.strip()
)

moms_notes['name_key'] = (
    moms_notes['first_name'].str.lower().str.strip() + '_' +
    moms_notes['last_name'].str.lower().str.strip()
)

# Step 2: Check how many will actually match BEFORE joining
master_keys = set(master['name_key'])
moms_keys   = set(moms_notes['name_key'])

matched    = master_keys & moms_keys          # intersection
only_master = master_keys - moms_keys         # in master but not moms
only_moms   = moms_keys - master_keys         # in moms but not master

print(f"✅ Will match:           {len(matched)}")
print(f"❌ Only in master:       {len(only_master)}")
print(f"❌ Only in moms_notes:   {len(only_moms)}")
```

```python
# Step 1: Do the left join on the name_key
master = master.merge(
    moms_notes,
    on='name_key',
    how='left',
    suffixes=('', '_moms')  # if columns have same name, keep master version
)

# Step 2: Drop the duplicate name columns moms_notes brought in
# (moms has first_name, last_name too — we already have those from identity)
master.drop(columns=['first_name_moms', 'last_name_moms', 'name_key'], 
            inplace=True, errors='ignore')

# Step 3: Drop the 100% useless distractor columns
cols_to_drop = [
    'eye_color',            # no hockey relevance
    'stuffed_animal_name',  # obvious noise
    'favourite_board_game', # obvious noise
    'favourite_child_info', # obvious noise
    'favourite_tv_show',    # 100% empty strings
    'favourite_food',       # 100% empty strings
    'physician_signature',  # joke values (Dr. Strange etc.)
]
master.drop(columns=cols_to_drop, inplace=True, errors='ignore')

# Step 4: Verify final shape
print(f"Final master shape: {master.shape}")
print(f"\nColumns remaining:")
print(list(master.columns))
```

```python
def parse_height_to_cm(value):
    """Convert height to centimeters regardless of input format."""
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip()
    
    if value.endswith('cm'):
        # e.g. "180.1135cm" → remove 'cm', convert to float
        return float(value[:-2])
    
    elif value.endswith('m'):
        # e.g. "1.79m" → remove 'm', convert to float, multiply by 100
        return float(value[:-1]) * 100
    
    else:
        # bare number with no unit — assume centimeters
        return float(value)

# Apply to the column
master['height'] = master['height'].apply(parse_height_to_cm)

print("Height after parsing (in cm):")
print(master['height'].describe())
```

```python
def parse_weight(value):
    """Strip 'kg' suffix if present, return float."""
    if pd.isna(value):
        return np.nan
    value = str(value).strip()
    if value.endswith('kg'):
        return float(value[:-2])
    else:
        return float(value)

def parse_body_fat(value):
    """Strip '%' suffix if present, return float."""
    if pd.isna(value):
        return np.nan
    value = str(value).strip()
    if value.endswith('%'):
        return float(value[:-1])
    else:
        return float(value)

master['weight']             = master['weight'].apply(parse_weight)
master['body_fat_percentage'] = master['body_fat_percentage'].apply(parse_body_fat)

print("Weight (kg):")
print(master['weight'].describe())
print("\nBody fat %:")
print(master['body_fat_percentage'].describe())
```

```python
# --- Numeric columns: impute with MEDIAN ---
# (Median is better than mean when we don't know if there are outliers)

median_contracts = master['contracts_signed'].median()
median_teams     = master['number_of_previous_teams'].median()

master['contracts_signed'].fillna(median_contracts, inplace=True)
master['number_of_previous_teams'].fillna(median_teams, inplace=True)

print(f"contracts_signed median used: {median_contracts}")
print(f"number_of_previous_teams median used: {median_teams}")

# --- Categorical column: dominant_hand → mode (most common value) ---
most_common_hand = master['dominant_hand'].mode()[0]
master['dominant_hand'].fillna(most_common_hand, inplace=True)

print(f"dominant_hand mode used: {most_common_hand}")

# --- Verify these 3 are now clean ---
print("\nMissing after imputation:")
print(master[['contracts_signed', 'number_of_previous_teams', 'dominant_hand']].isnull().sum())
```
