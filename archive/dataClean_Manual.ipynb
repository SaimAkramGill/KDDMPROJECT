#### Changing the roman names to english 
After checking the dataset finds out language difference and convert the columns to english like in the Identity Card 1 table. 

#### Also the first name and last names were in latin and finds out names ending with 'ius' and 'us'. So I have removed them and convert them to english

```python
translation_dict = {
    'numerus_internationalis_ad_identitatem': 'international_id',
    'numerus_identificationis_ad_medicinam': 'medical_id',
    'praenomen': 'first_name',
    'cognomen': 'last_name',
    'sexus': 'gender',
    'aetas_annorum': 'age',
    'urbs_natalis': 'birth_city',
    'natio': 'nationality'
}
identity_card_0.rename(columns=translation_dict, inplace=True)
```

```python
def roman_to_int(s):
    if pd.isna(s): return s
    roman_values = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    total = 0
    prev_value = 0
    
    for char in reversed(str(s).strip().upper()):
        if char not in roman_values: 
            return s 
        value = roman_values[char]
        if value < prev_value:
            total -= value
        else:
            total += value
        prev_value = value
    return total

identity_card_0['international_id'] = identity_card_0['international_id'].apply(roman_to_int)


def strip_latin_suffix(name):
    """
    Strip common Latin suffixes from Latinized names.
    Examples:
      'Alleeius' -> 'Allee'    (strip -ius)
      'Kohlius'  -> 'Kohl'     (strip -ius)
      'Keithus'  -> 'Keith'    (strip -us)
      'Donaldus' -> 'Donald'   (strip -us)
    """
    if pd.isna(name):
        return name
    name = str(name)
    if name.endswith('ius'):
        return name[:-3]  
    elif name.endswith('us'):
        return name[:-2]  
    return name  


identity_card_0['first_name'] = identity_card_0['first_name'].apply(strip_latin_suffix)
identity_card_0['last_name']  = identity_card_0['last_name'].apply(strip_latin_suffix)


print("After stripping Latin suffixes:")
display(identity_card_0[['first_name', 'last_name']].head(10))

print("\n--- Cleaned Identity Card 0 Data ---")
display(identity_card_0.head())
```

```python
def parse_height_to_cm(value):
    """Convert height to centimeters regardless of input format."""
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip()
    
    if value.endswith('cm'):
        # e.g. "180.1135cm" → remove 'cm', convert to float
        return float(value[:-2])
    
    elif value.endswith('m'):
        # e.g. "1.79m" → remove 'm', convert to float, multiply by 100
        return float(value[:-1]) * 100
    
    else:
        # bare number with no unit  assume centimeters
        return float(value)

# Apply to the column
master['height'] = master['height'].apply(parse_height_to_cm)

print("Height after parsing (in cm):")
print(master['height'].describe())
```

```python
def parse_weight(value):
    """Strip 'kg' suffix if present, return float."""
    if pd.isna(value):
        return np.nan
    value = str(value).strip()
    if value.endswith('kg'):
        return float(value[:-2])
    else:
        return float(value)

def parse_body_fat(value):
    """Strip '%' suffix if present, return float."""
    if pd.isna(value):
        return np.nan
    value = str(value).strip()
    if value.endswith('%'):
        return float(value[:-1])
    else:
        return float(value)

master['weight']             = master['weight'].apply(parse_weight)
master['body_fat_percentage'] = master['body_fat_percentage'].apply(parse_body_fat)

print("Weight (kg):")
print(master['weight'].describe())
print("\nBody fat %:")
print(master['body_fat_percentage'].describe())
```

```python
# A1  body_fat_percentage was never parsed; strip the '%' and coerce to float.
def parse_pct(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip().rstrip("%")
    try:
        return float(s)
    except ValueError:
        return np.nan

before = master["body_fat_percentage"].dtype
master["body_fat_percentage"] = master["body_fat_percentage"].map(parse_pct)
print(f"body_fat_percentage dtype: {before} -> {master['body_fat_percentage'].dtype}")
print(f"range: {master['body_fat_percentage'].min():.2f} – {master['body_fat_percentage'].max():.2f}")
```

```python
cols_to_check = ["shooting_percentage", "salary", "time_on_ice", "sprint_time", "weight"]
cols_to_check = [c for c in cols_to_check if c in master.columns]

fig, axes = plt.subplots(1, len(cols_to_check), figsize=(3.2 * len(cols_to_check), 4.5))
if len(cols_to_check) == 1:
    axes = [axes]

for ax, col in zip(axes, cols_to_check):
    data = master[col].dropna()
    ax.boxplot(data, vert=True, patch_artist=True,
               boxprops=dict(facecolor="#0B2E4F", color="#0B2E4F"),
               medianprops=dict(color="#C8102E", linewidth=2),
               flierprops=dict(marker="o", markersize=3,
                               markerfacecolor="#C8102E", markeredgecolor="none"))
    ax.set_title(col, fontsize=10)
    q1, q3 = data.quantile([0.25, 0.75])
    iqr = q3 - q1
    lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_out = ((data < lo) | (data > hi)).sum()
    ax.set_xlabel(f"n_outliers={n_out}")
    ax.set_xticks([])

plt.suptitle("IQR outlier scan (1.5×IQR rule)", fontweight="bold")
plt.tight_layout()
plt.show()
```

### F. EDA  Derived feature consistency

`pass_completion_rate` should equal `passes_completed / passes_attempted × 100`. If it does, the file is internally coherent  a mark of trustworthy data.

```python
mask = master["passes_attempted"] > 0
recomputed = (
    master.loc[mask, "passes_completed"]
    / master.loc[mask, "passes_attempted"] * 100
).round(2)
diff = (recomputed - master.loc[mask, "pass_completion_rate"]).abs()

print(f"rows checked: {mask.sum()}")
print(f"rows where reported rate differs from recomputed by > 1pp: {(diff > 1).sum()}")
print(f"max absolute deviation: {diff.max():.4f} pp")
```

### G. Imputation strategy



```python
# Numeric: median by position (falls back to global median when a position group is fully null).
numeric_by_position = ["weight", "body_fat_percentage", "sprint_time", "shoe_size", "height"]
for col in numeric_by_position:
    if col not in master.columns:
        continue
    fill = master.groupby("position")[col].transform("median")
    master[col] = master[col].fillna(fill).fillna(master[col].median())

# Numeric: column median for the small-null columns we've already touched.
small_numeric = ["age", "draft_year", "salary", "years_played", "years_pro",
                 "school_grade", "years_in_usa"]
for col in small_numeric:
    if col not in master.columns:
        continue
    master[col] = master[col].fillna(master[col].median())

# Categorical: mode by position for fitness_level.
if "fitness_level" in master.columns:
    fill = master.groupby("position")["fitness_level"].transform(
        lambda s: s.mode().iloc[0] if not s.mode().empty else np.nan
    )
    master["fitness_level"] = master["fitness_level"].fillna(fill)

# Verify what's left.
remaining = master.isnull().sum()
remaining = remaining[remaining > 0].sort_values(ascending=False)
print("columns still containing NaN (intentionally left for Phase II):")
print(remaining)
```
